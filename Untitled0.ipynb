```python
import csv


def load_data(filename):
    """Read the CSV and return (attribute_names, rows) without the Example column."""
    with open(filename, newline="") as f:
        reader = csv.reader(f)
        header = next(reader)[1:]          # skip 'Example' column
        rows = [row[1:] for row in reader]
    return header, rows


def find_s(rows, n_attrs):
    """FIND-S: returns the most specific hypothesis consistent with positive examples."""
    # Step 1: initialise h to the most specific hypothesis
    hypothesis = ["0"] * n_attrs
    print(f"Initial hypothesis : {hypothesis}\n")

    for i, row in enumerate(rows, start=1):
        attributes, target = row[:-1], row[-1]

        # Step 2: ignore negative examples
        if target.strip().lower() != "yes":
            print(f"Example {i}: {attributes} -> {target} (negative, ignored)")
            print(f"           h = {hypothesis}\n")
            continue

        # Step 3: generalise h only where it disagrees with a positive example
        for j in range(n_attrs):
            if hypothesis[j] == "0":
                hypothesis[j] = attributes[j]       # first positive example
            elif hypothesis[j] != attributes[j]:
                hypothesis[j] = "?"                 # generalise

        print(f"Example {i}: {attributes} -> {target} (positive)")
        print(f"           h = {hypothesis}\n")

    return hypothesis


if __name__ == "__main__":
    attrs, data = load_data("s algorithm.csv")
    print("Attributes:", attrs[:-1], "\n")
    result = find_s(data, len(attrs) - 1)
    print("Final most specific hypothesis:", result)
```
